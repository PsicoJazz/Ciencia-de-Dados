```python
import torch
import torch.nn as nn
import pandas as pd
```

```python
def celcius_to_fahrenheit(celcius):
    return (1.8 * celcius + 32)
```

### Conversão de 60 graus Celsius para Fahrenheit

```python
temp_celcius = [-10, 20, 100]
temp_fare = [celcius_to_fahrenheit(temp) for temp in temp_celcius]
```

```python
df = pd.DataFrame({'celcius': temp_celcius, 'fare': temp_fare})
df
```

```python
x = torch.FloatTensor(df['celcius'].values).unsqueeze(1) # Garante que x tenha a forma (batch_size, 1)
y = torch.FloatTensor(df['fare'].values).unsqueeze(1)
```

```python
class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.input_layer = nn.Linear(in_features=1, out_features=1, bias=True)

    def forward(self, x):
        out = self.input_layer(x)
        return out
```

```python
EPOCHS = 1000
LR = 0.2
```

```python
model = Model()
criterion = nn.MSELoss()
optmizer = torch.optim.Adam(params=model.parameters(), lr=LR)
```

```python
weights = []
bias = []

for epoch in range(EPOCHS):
  output = model.forward(x)
  loss = criterion(output, y)
  optmizer.zero_grad()
  loss.backward()
  optmizer.step()

  weights.append(model.input_layer.weight.item())
  bias.append(model.input_layer.bias.item())

  print(f'Peso: {model.input_layer.weight.item():.2f}')
  print(f'Bias: {model.input_layer.bias.item():.2f}')
```

```python
training_data = pd.DataFrame({'Pesos': weights, 'Bias':bias})
training_data.to_csv('training_data.csv', index=False)
```
